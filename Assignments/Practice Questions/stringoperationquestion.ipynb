```python
text = input("Enter a string: ")
count = 0
vowels = {'a':0, 'e':0, 'i':0, 'o':0, 'u':0, }
for ch in text:
    if ch in "aeiou":
        count += 1
        vowels[ch.lower()] += 1
print("Number of vowels: ",count)
print("Vowel Counts: ",vowels)
```

```python
s = input("Enter a string: ")

rev = []

for i in range(len(s)-1, -1, -1):
    rev.append(s[i])

rev_str = ""

for ch in rev:
    rev_str += ch

if s == rev_str:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
s = input("Enter a string: ")

freq = {}

for ch in s:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

print(freq)
```

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

short_words = []
medium_words = []
long_words = []

for word in words:
    if len(word) < 5:
        short_words.append(word)
    elif len(word) <= 8:
        medium_words.append(word)
    else:
        long_words.append(word)

print("Short words:", short_words)
print("Medium words:", medium_words)
print("Long words:", long_words)
```

```python
strings = ["python", "pythons", "pyon", "py123thon", "java"]

matching_strings = []

for s in strings:
    if s.startswith("py") and s.endswith("thon"):
        matching_strings.append(s)

print(matching_strings)
```

```python
paragraph = input("Enter a paragraph: ")

paragraph = paragraph.lower()

for ch in ",.!?;:":
    paragraph = paragraph.replace(ch, "")

words = paragraph.split()

unique_words = set()

for word in words:
    unique_words.add(word)

result = sorted(list(unique_words))

print(result)
```

```python
s = input("Enter a string: ")

even_chars = []
odd_chars = []

for index, ch in enumerate(s):
    if index % 2 == 0:
        even_chars.append(ch)
    else:
        odd_chars.append(ch)

print("Even index characters:", even_chars)
print("Odd index characters:", odd_chars)
```

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

list1 = list(s1)
list2 = list(s2)

list1.sort()
list2.sort()

if list1 == list2:
    print("Anagrams")
else:
    print("Not anagrams")
```

```python
s = input("Enter a string: ").lower()

letters = set("abcdefghijklmnopqrstuvwxyz")

for ch in s:
    if ch in letters:
        letters.remove(ch)

print("Missing letters:", letters)
```

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

reversed_words = []

for i in range(len(words)-1, -1, -1):
    reversed_words.append(words[i])

result = " ".join(reversed_words)

print(result)
```
