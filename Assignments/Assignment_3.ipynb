```python
class BankAccount:
    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance -= amount

    def get_balance(self):
        return self.balance

    def __str__(self):
        return f"Account Holder: {self.account_holder}\nAccount Number: {self.account_number}\nBalance: {self.balance}"


acc1 = BankAccount("Shrushti", 101, 5000)
acc2 = BankAccount("Riya", 102, 6000)

acc1.deposit(6000)
acc1.withdraw(2000)

acc2.deposit(7000)
acc2.withdraw(10000)

print(acc1)
print()
print(acc2)
```

```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def calculate_percentage(self):
        return sum(self.marks) / len(self.marks)

    def get_grade(self):
        per = self.calculate_percentage()

        if per >= 80:
            return "A"
        elif per >= 60:
            return "B"
        elif per >= 40:
            return "C"
        else:
            return "Fail"

    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.roll_no)
        print("Marks:", self.marks)
        print("Percentage:", self.calculate_percentage())
        print("Grade:", self.get_grade())
        print()


s1 = Student("Amit", 1, [80, 85, 90, 75, 88])
s2 = Student("Riya", 2, [70, 65, 60, 75, 80])
s3 = Student("Rahul", 3, [35, 40, 45, 30, 50])

students = [s1, s2, s3]

for i in students:
    i.display()
```

```python
class Employee:
    def __init__(self, name, emp_id, salary):
        self.__name = name
        self.__emp_id = emp_id
        self.set_salary(salary)

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_emp_id(self):
        return self.__emp_id

    def set_emp_id(self, emp_id):
        self.__emp_id = emp_id

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 10000:
            raise ValueError("Salary must be at least 10000")
        self.__salary = salary

    def give_raise(self, percent):
        self.__salary += self.__salary * percent / 100


emp = Employee("Shrushti", 101, 20000)

print(emp.get_name())
print(emp.get_salary())

emp.give_raise(10)

print("New Salary:", emp.get_salary())
```

```python
class SecureVault:
    def __init__(self, pin):
        self.__pin = pin
        self.__data = {}

    def add_entry(self, pin, key, value):
        if pin == self.__pin:
            self.__data[key] = value
        else:
            print("Access Denied")

    def get_entry(self, pin, key):
        if pin == self.__pin:
            return self.__data.get(key)
        else:
            print("Access Denied")


vault = SecureVault("1234")

vault.add_entry("1234", "Email", "abc@gmail.com")
print(vault.get_entry("1234", "Email"))

vault.get_entry("1111", "Email")
```

```python
import math

class Shape:
    def area(self):
        return 0

    def describe(self):
        print("I am a shape")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius

    def describe(self):
        print("I am a Circle")


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def describe(self):
        print("I am a Rectangle")


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

    def describe(self):
        print("I am a Triangle")


c = Circle(5)
r = Rectangle(4, 6)
t = Triangle(10, 8)

c.describe()
print("Area:", c.area())

r.describe()
print("Area:", r.area())

t.describe()
print("Area:", t.area())
```

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def start_engine(self):
        print("Engine started")


class Car(Vehicle):
    def __init__(self, make, model, year, num_doors):
        super().__init__(make, model, year)
        self.num_doors = num_doors

    def start_engine(self):
        print(f"Engine started for {self.make} {self.model}")


class ElectricCar(Vehicle):
    def __init__(self, make, model, year, battery_capacity):
        super().__init__(make, model, year)
        self.battery_capacity = battery_capacity

    def start_engine(self):
        print(f"Silent start - battery at {self.battery_capacity} kWh")


c1 = Car("Honda", "City", 2023, 4)
e1 = ElectricCar("Tesla", "Model 3", 2024, 75)

c1.start_engine()
e1.start_engine()
```

```python
class Animal:
    def sound(self):
        print("Some sound")


class Dog(Animal):
    def sound(self):
        print("Bark")


class Cat(Animal):
    def sound(self):
        print("Meow")


class Cow(Animal):
    def sound(self):
        print("Moo")


def make_all_sounds(animals):
    for animal in animals:
        animal.sound()


animals = [Dog(), Cat(), Cow()]

make_all_sounds(animals)
```

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

    def multiply(self, a, b, c=1):
        return a * b * c


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))

print(calc.multiply(2, 3))
print(calc.multiply(2, 3, 4))


class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"


v1 = Vector(2, 3)
v2 = Vector(4, 5)

v3 = v1 + v2

print(v3)
```

```python
class StudentManager:
    def __init__(self):
        self.students = []

    def add_student(self):
        name = input("Enter Name: ")
        roll = input("Enter Roll No: ")

        marks = []
        for i in range(3):
            m = float(input(f"Enter Marks {i+1}: "))
            marks.append(m)

        self.students.append([name, roll, marks])

    def view_all(self):
        for s in self.students:
            avg = sum(s[2]) / len(s[2])
            print("Name:", s[0])
            print("Roll No:", s[1])
            print("Average:", avg)
            print()

    def search_student(self):
        roll = input("Enter Roll No: ")

        for s in self.students:
            if s[1] == roll:
                print("Name:", s[0])
                print("Marks:", s[2])
                return

        print("Student Not Found")

    def delete_student(self):
        roll = input("Enter Roll No: ")

        for s in self.students:
            if s[1] == roll:
                self.students.remove(s)
                print("Student Deleted")
                return

        print("Student Not Found")

    def save_to_file(self):
        f = open("students_data.txt", "w")

        for s in self.students:
            f.write(f"{s[0]},{s[1]},{s[2]}\n")

        f.close()
        print("Data Saved")

    def load_from_file(self):
        try:
            f = open("students_data.txt", "r")

            for line in f:
                print(line.strip())

            f.close()

        except:
            print("No file found")


manager = StudentManager()

while True:
    print("\n1. Add Student")
    print("2. View All")
    print("3. Search Student")
    print("4. Delete Student")
    print("5. Save to File")
    print("6. Load from File")
    print("7. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        manager.add_student()

    elif choice == 2:
        manager.view_all()

    elif choice == 3:
        manager.search_student()

    elif choice == 4:
        manager.delete_student()

    elif choice == 5:
        manager.save_to_file()

    elif choice == 6:
        manager.load_from_file()

    elif choice == 7:
        print("Exiting...")
        break

    else:
        print("Invalid Choice")
```
