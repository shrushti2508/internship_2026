```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
add = num1 + num2
print("Sum: ",add)
diff = num1 - num2
print("Difference: ",diff)
mul = num1 * num2
print("Product: ",mul)
div = num1 / num2
if num2 != 0:
    print("Quotient: ",div)
else:
    print("Cannot divide by zero")
```

```python
celsius = float(input("Enter the temperature in celsius: "))
fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15
print("Celsius: ",celsius)
print("Fahrenheit: ",fahrenheit)
print("Kelvin: ",kelvin)
```

```python
year = int(input("Enter the year: "))
if year % 10 == 0:
    if year % 400 == 0:
        print(f"{year} is century year and leap year")
    else:
        print(f"{year} is century year but not leap year")
else:
    if year % 4 == 0:
        print(f"{year} is a leap year")
    else:
        print(f"{year} is not a leap year")
```

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
print("Menu: \n 1. + operator \n 2. - operator \n 3. * operator \n 4. / operator")
ch = int(input("Enter the choice: "))
if ch == 1:
    print("Addition: ",num1 + num2)
elif ch == 2:
    print("Subtraction: ",num1 - num2)
elif ch == 3:
    print("Multiplication: ",num1 * num2)
elif ch == 4:
    if num2 != 0:
        print("Diviswion: ",num1 / num2)
    else:
        print("Cannot divide by zero")
else:
    print("Invalid operator")
```

```python
n = int(input("Enter the number: "))
for num in range(2,n+1):
    is_prime = True
    for i in range(2,num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, end=" ")
```

```python
num = int(input("Enter the number until user enters 0: "))
count = 0
total = 0
while num != 0:
    count += 1
    total = total + num
    num = int(input("Enter the number until user enters 0: "))
if count > 0:
    print("Count: ",count)
    print("Sum: ",total)
    print("Average: ",total / count)
    
```

```python
num = int(input("Enter number of rows: "))
for i in range(1,num + 1):
    spaces = n - i
    stars = 2 * i - 1
    print(" " * spaces + "*" * stars)
```
