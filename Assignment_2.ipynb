```python
def analyze_sentence(sentence) :
    words = sentence.lower().split()
    
    result ={
        "word_count": len(words),
        "char_count": len(sentence),
        "unique_words": len(set(words)),
        "most_frequent_word": max(set(words), key=words.count)
    }

    print(result)
```

```python
analyze_sentence("The quick brown fox jumps over the lazy dog the fox")
```

```python
text = "A man a plan a canal Panama"

clear = text.replace(" ","").lower()

rev =""
for i in clear:
    rev = i + rev

if clear == rev:
    print("The string is palindrome ")
else:
    print("The string is not palindrome")
```

```python
text = "A man a plan a canal Panama"

new_str = ""

vowels = {"a","e" ,"i","o" ,"u"}

for i in text:
    if i.lower() in vowels:
        new_str +="*"
    else:
        new_str += i

print("Replace :", new_str)
```

```python
def format_report(data):

    print(f"{"Name":<15}{"Score":>6}{"Grade":^8}")
    print("-" * 29)

    for name , score , grade in data:
        print(f"{name:<15}{score:>6}{grade:^8}")
```

```python
format_report( [("Alice", 92, "A"), ("Bob", 78, "B+"), ("Charlie", 55, "C"), ("Diana",
100, "A+")])
```

```python
def list_ops(nums):
      
    nd = []
    for i in nums:
        if i not in nd :
            nd.append(i)
    print("Non duplicate numbers:" , nd)


    es = []
    for i in range(0,len(nums),2):
        es.append(nums[i] **2)
    print("even indexed square:" ,es)
    
    nested = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]
    flat = [num for sub in nested for num in sub]
    print("Flattened list :", flat)
```

```python
list_ops([3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5])
```

```python
def inventory_manager(records):
    result ={}
    for product , category , quantity , price in records:
        if category not in result :
            result[category] ={
                "total_items" : 0 , 
                
                "total_value" : 0 ,
                    
                "product" : []
            }
        result[category]["total_items"] +=quantity
        
        result[category]["total_value"] +=quantity * price
        
        result[category]["product"].append(product)

    return result
```

```python
inventory_manager([ 
("Laptop", "Electronics", 10, 55000),
("Mouse", "Electronics", 50, 499), 
("Desk", "Furniture", 8, 12000), 
("Chair", "Furniture", 15, 4500),
("Pen", "Stationery", 200, 20), ])
```

```python
def power_all(*args, exp=2):
    result =[]
    for i in args:
        result.append(i ** exp)
    return result 
```

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print("Execution time:", end - start, "seconds")
        return result
    return wrapper
```

```python
@timer
def slow_sum(n):
    return sum(i * i for i in range(1, n + 1))


print(slow_sum(100000))
```

```python
def validate_positive(func):
    def wrapper(x):
        if x <= 0:
            raise ValueError("Input must be positive")
        return func(x)
    return wrapper
```

```python
import math

@validate_positive
def safe_sqrt(x):
    return math.sqrt(x)


print(safe_sqrt(25))  
```
