```python
numbers = [12, 7, 23, 48, 15, 66, 9]

sum_even = 0

for i in numbers:
    if i % 2 == 0:
        sum_even += i

print("Sum of even numbers =", sum_even)
```

```python
data = ['apple', 'banana', 'cherry', 'date', 'elderberry']

reverse_list = []

for i in range(len(data)-1, -1, -1):
    reverse_list.append(data[i])

print(reverse_list)
```

```python
values = [8, 2, 10, 5, 17, 3, 14]

largest = values[0]

for i in values:
    if i > largest:
        largest = i

print("Largest number =", largest)
```

```python
items = ['red', 'blue', 'green', 'red', 'yellow', 'blue', 'red']

count = 0

for i in items:
    if i == 'red':
        count += 1

print("Count =", count)
```

```python
mixed_numbers = [1, 5, 2, 5, 3, 5, 4, 5]

new_list = []

for i in mixed_numbers:
    if i != 5:
        new_list.append(i)

print(new_list)
```

```python
list1 = [1, 2, 3, 2, 1]

reverse_list = []

for i in range(len(list1)-1, -1, -1):
    reverse_list.append(list1[i])

if list1 == reverse_list:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
list2 = ['a', 'b', 'c', 'b', 'd']

reverse_list = []

for i in range(len(list2)-1, -1, -1):
    reverse_list.append(list2[i])

if list2 == reverse_list:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
list_a = [3, 1, 4]
list_b = [2, 5, 0]

combined = list_a + list_b

for i in range(len(combined)):
    for j in range(len(combined)-1):
        if combined[j] > combined[j+1]:
            combined[j], combined[j+1] = combined[j+1], combined[j]

print(combined)
```

```python
nested_list = [1, [2, 3], 4, [5, 6, 7], 8]

flat_list = []

for i in nested_list:
    if type(i) == list:
        for j in i:
            flat_list.append(j)
    else:
        flat_list.append(i)

print(flat_list)
```
