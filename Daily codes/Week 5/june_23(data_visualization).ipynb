```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.array([10,23,29,34,39,45,50])
y = np.array([14,21,31,43,48,55,58])
z = np.array([56,50,44,37,26,20,16])
```

## Line Chart

```python
plt.plot(x,y);
```

```python
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Line Chart: x vs y")
plt.plot(x,y, marker = 'o', mfc = 'red', ls = '--');
plt.grid()
```

```python
plt.xlabel("x values")
plt.ylabel("z values")
plt.title("Line Chart: x vs z")
plt.plot(x,z, marker = 'o', mfc = 'orange', ls = '--');
plt.grid()
```

```python
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Line Chart: x vs y")
plt.plot(x,y, marker = 'o', mfc = 'red', ls = '--');
plt.plot(x,z, marker = 'o', mfc = 'black', ls = '--', color = 'orange');
plt.grid()
```

```python
plt.figure(figsize = (16, 9))

plt.subplot(2,2,1)
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Line Chart: x vs y")
plt.plot(x,y, marker = 'o', mfc = 'red', ls = '--');
plt.grid()

plt.subplot(2,2,2)
plt.xlabel("x values")
plt.ylabel("z values")
plt.title("Line Chart: x vs z")
plt.plot(x,z, marker = 'o', mfc = 'red', ls = '--');
plt.grid()
```

## Bar Chart

```python
courses = ['python', 'lisp', 'c', 'R', 'php']
students = [68, 30, 47, 39, 60]
```

```python
plt.title("Student Enrollment")
plt.xlabel("course")
plt.ylabel("no.of students enrolled")
plt.bar(courses, students, color = 'cyan', edgecolor = 'red');
plt.grid();
```
