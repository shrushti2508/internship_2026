```python
string = input("Enter a string: ")
char1 = input("Enter the first character: ")
char2 = input("Enter the second character: ")
found_char1 = False
found_char2 = False
for char in string:
    if char == char1:
        found_char1 = True
    if char == char2:
        found_char2 = True
if found_char1:
    print(f"The character '{char1}' is present in the string")
else:
    print(f"The character '{char1}' is not present in the string")
if found_char2:
    print(f"The character '{char2}' is present in the string")
else:
    print(f"The character '{char2}' is not present in the string")
```
