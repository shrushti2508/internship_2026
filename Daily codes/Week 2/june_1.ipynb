```python
names = ["Aryan","atharva","samruddhi","swaraj","sukanya","Shivraj"]
new3 = [i for i in names if i.startswith(('s','S'))]
new3
```

```python
new4 = [i for i in names if i.lower().startswith('s')]
new4
```

```python
new5 = [i for i in names if i[0] == 's' or i[0] == 'S']
new5
```
