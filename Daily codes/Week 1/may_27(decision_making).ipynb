```python
purchase_cost = float(input("Enter the purchase cost: "))
selling_cost = float(input("Enter the selling cost:"))
if selling_cost > purchase_cost:
    profit = selling_cost - purchase_cost
    print("Profit =", profit)

elif purchase_cost > selling_cost:
    loss = purchase_cost - selling_cost
    print("Loss =", loss)

else:
    print("No Profit No Loss")
```

```python
#if the year is divided by 100 and 400 then it is century leap year

year = int(input("Enter the year: "))
if year % 100 == 0:         
    if year % 400 == 0:
        print("Century Leap Year")
    else:
        print("Century but Not a Leap Year") 

else:                        
    if year % 4 == 0:
        print("Leap Year")
    else:
        print("Not a Leap Year")
```

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
num3 = int(input("Enter the third number: "))
if num1 > num2:
    if num1 > num3:
        print(num1, "is largest")
    else:
        print(num3, "is largest")
else:
    if num2 > num3:
        print(num2, "is largest")
    else:
        print(num3, "is largest")
```

```python
print("Welcome!")
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
print("Menu: \n 1. Addition \n 2. Subtraction \n 3. Multiplication \n 4. Division")
ch = int(input("Select choice: "))
if ch == 1:
    print("Addition: ",num1 + num2)
elif ch == 2:
    print("Subtraction: ",num1 - num2)
elif ch == 3:
    print("Multiplication: ",num1 * num2)
elif ch == 4:
    print("Division: ",num1 / num2)
else:
    print("Invalid choice")
```

```python
str = input("Enter the string: ")
print("Menu: \n 1. Convert to uppercase \n 2. Convert to lowercase \n 3. Replace a particular character")
ch = int(input("Select choice: "))
if ch == 1:
    print("Uppercase:", str.upper())
elif ch == 2:
    print("Lowercase:", str.lower())
elif ch == 3:
    character = input("Enter the character by which it should be replaced: ")
    pos = (input("Enter the position of the character to be replaced: "))
    print("After replacing: ",str.replace(pos,character))
else:
    print("Invalid choice")
    
    
```
