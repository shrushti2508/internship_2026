```python
try:
    f = open('a1.txt')
    print(var1)
except FileNotFoundError:
    print("file does not exist")
except Exception as e:
    print(e)
```

```python
try:
    f = open('a1.txt')
except FileNotFoundError:
    print("File does not exist")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()
```

```python
#finally block
```

```python
try:
    f = open('a1.txt')
    print(var1)
except FileNotFoundError:
    print("File does not exist")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()
finally:
    print("Run complete")
```

```python
try:
    f = open('a1.txt')
    if f.name == 'a1.txt':
        raise Exception
except FileNotFoundError:
    print("File does not exist")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()
finally:
    print("Run complete")
```

```python
f = open('a1.txt')
if f.name == 'a1.txt':
    raise Exception
```

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
try:
    c = a / b
    print("Division: ",c)
except ZeroDivisionError:
    print("error: division by zero")
except ValueError:
    print("error: enter a valid number")
```
