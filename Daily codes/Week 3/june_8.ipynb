```python
class Person:

    def __init__(self, name, age, ht):
        self.name = name
        self.__age = age #private variable
        self.height = ht

    def show(self):
        return f"Name : {self.name}, Age: {self.__age}, Height: {self.height}"
```

```python
p1 = Person('Jatin', 20, 5.8)
p2 = Person('Sachin', 22, 5.9)
```

```python
p1.show()
```

```python
class Employee:
    company = "TCS"

    def __init__(self, name, monthly_salary, job_designation):
        self.name = name
        self.__monthly_salary = monthly_salary #private variable
        self.job_designation = job_designation

    def annual_salary(self):
        return self.__monthly_salary * 12

    def show_data(self):
        print("Company: ", Employee.company)
        print("Name: ", self.name)
        print("Job Designation: ", self.job_designation)
        print("Monthly Salary: ", self.__monthly_salary)
        print("Annual Salary: ", self.annual_salary())
        print("  ")

e1 = Employee("Shrushti", 30000, "Python Developer")
e2 = Employee("Omkar", 35000, "Data Analyst")
e3 = Employee("Riya", 40000, "Software Engineer")

e1.show_data()
e2.show_data()
e3.show_data()
```
