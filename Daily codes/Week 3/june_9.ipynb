## Inheritance

```python
class Person:
    insti = "MITU"

    def __init__(self, age, ht):
        self.age = age
        self.ht = ht

    def show(self):
        print(f"Age: {self.age}, Height: {self.ht}, Institute: {self.insti}")
```

```python
class Ravi(Person):
    def __init__(self, age, ht, sal):
        super().__init__(age, ht)
        self.sal = sal

    def show_sal(self):
        print(f"Salary: {self.sal} ")
```

```python
a = Ravi(25, 5.9, 57000)
```

```python
a.show_sal()
```

```python
a.show()
```

```python
class Menuapp:
    def __init__(self):
        self.itemlist = []
    def show_menu(self):
        print("Menu \n 1. Add items \n 2. View items \n 3. Exit")
    def add_items(self):
        item = input("enter the item name: ")
        self.itemlist.append(item)
    def view_items(self):
        return self.itemlist
    def run(self):
        while True:
            self.show_menu()
            ch = int(input("Enter the choice: "))
            if ch == 1:
                self.add_items()
            elif ch == 2:
                self.view_items()
            elif ch == 3:
                break
            else:
                return "Invalid choice"
                
```

```python
a = Menuapp()
```

```python
a.run()
```

```python
a.itemlist
```

```python
class Person:
    def __init__(self, age, name):
        self.__age = age
        self.__name = name

    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

    def set_age(self, age):
        self.__age = age

    def __str__(self):
        return f"Age: {self.__age}, Name: {self.__name}"
```

```python
p = Person(34, 'Arun')
```

```python
print(p)
```

```python
class BankAccount:

    def __init__(self):
        self.balance = 1000

    def check_balance(self):
        print("Current Balance =", self.balance)

    def deposit(self):
        amount = float(input("Enter amount to deposit: "))
        self.balance += amount
        print("Amount deposited successfully.")

    def withdraw(self):
        amount = float(input("Enter amount to withdraw: "))

        if amount <= self.balance:
            self.balance -= amount
            print("Amount withdrawn successfully.")
        else:
            print("Insufficient balance!")

account = BankAccount()

while True:
    print("\n Banking Menu")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        account.check_balance()

    elif choice == 2:
        account.deposit()

    elif choice == 3:
        account.withdraw()

    elif choice == 4:
        print("Thank you for using the bank app.")
        break

    else:
        print("Invalid choice!")
```
